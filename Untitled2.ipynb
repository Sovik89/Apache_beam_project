```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder.enableHiveSupport().appName("Test").getOrCreate()
```

```python
data = [ (1, "Laptop", 10, 50000, "North", "2025-01-01"), (2, "Mobile", None, 15000, "South", None), (3, "Tablet", 20, None, "West", "2025-01-03"), (4, "Desktop", 15, 30000, None, "2025-01-04"), (5, None, None, None, "East", "2025-01-05") ]
columns = ["Sales_ID", "Product", "Quantity", "Price", "Region", "Sales_Date"]
```

```python
df = spark.createDataFrame(data=data,schema=["Sales_ID", "Product", "Quantity", "Price", "Region", "Sales_Date"])
```

```python
df.printSchema()
```

```python
df.show()
```

```python
#e. Drop rows where all columns are NULL
from functools import reduce

df=df.withColumn("All_values_null",reduce(lambda a, b: a & b, (col(c).isNull() for c in df.columns)))
```

```python
df=df.filter(col("All_values_null")==False).drop(col("All_values_null"))
df.show()
```

```python
from pyspark.sql.functions import col, when, lit

# Step 1: Create a new column replacing nulls with 0
df = df.withColumn("NewQuantity", when(col("Quantity").isNull(), lit(0)).otherwise(col("Quantity"))).\
                                                                                   drop("Quantity").\
                                                                                   withColumnRenamed("NewQuantity", "Quantity")
```

```python
df.show()
```

```python
avg_price=df.agg(round(avg(col("Price")),2).alias("avg_val")).collect()
```

```python
avg_price_val=avg_price[0]["avg_val"]
```

```python
avg_price_val
```

```python
# Step 2: Replace all NULL values in the Price column with the average price of the existing data
df = df.withColumn("NewPrice", when(col("Price").isNull(), lit(avg_price_val)).otherwise(col("Price"))).\
                                                                                  drop("Price").\
                                                                                   withColumnRenamed("NewPrice", "Price")
```

```python
df.show()
```

```python
#c. Drop rows where the Product column is NULL.

df=df.where("Product is not NULL")
```

```python
df.show()
```

```python
#d. Fill missing Sales_Date with a default value of '2025-01-01'

df = df.withColumn("NewSales_Date", when(col("Sales_Date").isNull(),'2025-01-01' ).otherwise(col("Sales_Date"))).\
                                                                                  drop("Sales_Date").\
                                                                                   withColumnRenamed("NewSales_Date", "Sales_Date")
```
